# Introduction to NumPy

### What is NumPy?
NumPy (Numerical Python) is an open-source Python library used for performing mathematical and numerical operations efficiently. It provides a powerful object called ndarray (N-dimensional array) that stores data in a compact and optimized format.

NumPy is the foundation of many AI and Machine Learning libraries such as:

* Pandas
* Scikit-learn
* TensorFlow
* PyTorch
* OpenCV

```python
import numpy as np

arr = np.array([10,20,30,40])
print(arr)
```

#### AI/ML Usage

Almost every Machine Learning dataset is converted into NumPy arrays before training a model because mathematical calculations are much faster.

### Why NumPy?

Python lists are useful, but they become slow when working with large datasets.

NumPy provides:

* Faster computation
* Less memory usage
* Built-in mathematical functions
* Easy matrix operations
* Supports multidimensional arrays

```python
import numpy as np

python_list = [1,2,3,4]

numpy_array = np.array([1,2,3,4])

print(type(python_list))
print(type(numpy_array))
```

#### AI/ML Usage

Machine learning algorithms perform thousands of mathematical operations.

Using NumPy makes these operations much faster than Python lists.

### NumPy Arrays (ndarray)

The main object in NumPy is called ndarray (N-dimensional array).

It stores data of the same data type in one place.

Unlike Python lists, NumPy arrays are homogeneous.

```python
import numpy as np

arr = np.array([5,10,15,20])
print(arr)
```

### Dimensions

Dimension means the number of axes in an array.

* [1,2,3]        ->	1D
* [[1,2],[3,4]]	 -> 2D
* [[[1]]]        ->	3D

```python
import numpy as np

a = np.array([1,2,3])

b = np.array([[1,2],[3,4]])

c = np.array([[[1],[2]],[[3],[4]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
```

#### AI/ML Usage

* Images → 3D arrays
* Videos → 4D arrays
* Deep Learning batches → 4D or 5D arrays

### Flatten

Converts multidimensional arrays into one dimension.

```python
arr = np.array([[1,2],
                [3,4]])

flat = arr.flatten()
print(flat)
```

#### AI/ML Usage

### Copy vs View
#### Copy

Creates a completely separate array.

```python
arr = np.array([1,2,3])

copy = arr.copy()

copy[0]=100

print(arr)
print(copy)
```

#### View

Shares the same memory.

```python
arr = np.array([1,2,3])

view = arr.view()

view[0]=100

print(arr)
print(view)
```

#### AI/ML Usage

Copies protect original datasets while preprocessing.

Views save memory for large datasets.

### Broadcasting

Broadcasting allows arrays with different shapes to perform arithmetic operations automatically.

```python
matrix = np.array([[1,2,3],
                   [4,5,6]])

vector = np.array([10,20,30])

print(matrix + vector)
```

#### AI/ML Usage

Feature scaling

New Value = Original Value - Mean

Mean values are broadcast across every row.

### Array Operations

Basic arithmetic operations.

```python
a = np.array([1,2,3])

b = np.array([4,5,6])

print(a+b)

print(a-b)

print(a*b)

print(a/b)
```

Some other Operations

```python
print(np.sum(a))

print(np.mean(a))

print(np.max(a))

print(np.min(a))
```

#### AI/ML Usage

Used while calculating:

* Mean

### Universal Functions (ufuncs)

Universal functions are built-in NumPy functions that operate element-wise on arrays.

They are fast because they are implemented in optimized C code.

```python
arr = np.array([1,4,9,16])

print(np.sqrt(arr))
```

Other ufuncs

```python
arr = np.array([1,2,3])

print(np.square(arr))

print(np.exp(arr))

print(np.log(arr))

print(np.sin(arr))

print(np.cos(arr))

print(np.abs([-2,-5,6]))
```

#### AI/ML Usage

Ufuncs are commonly used for:

* Feature scaling
* Data normalization
* Activation functions
* Statistical calculations
* Distance computations
