### Grouping & Aggregation

```python
import pandas as pd
```

```python
students = pd.DataFrame({
    "Student_ID": [101,102,103,104,105,106,107,108],
    "Name": ["Aishu","Rahul","Priya","John","Meena","David","Sneha","Arun"],
    "Department": ["AI","ML","AI","DS","ML","DS","AI","ML"],
    "Gender": ["Female","Male","Female","Male","Female","Male","Female","Male"],
    "Marks": [95,88,91,84,89,76,98,92],
    "Age": [22,23,21,24,22,23,21,22]
})

print(students)
```

### 1. groupby()

groupby() divides data into groups based on one or more columns and performs operations on each group independently.

It is similar to SQL's GROUP BY clause.

Syntax:

df.groupby("column")

Example:

A company calculates the average salary for each department.

#### AI/ML Usage

Used to summarize data before feature engineering and exploratory data analysis (EDA).

```python
students.groupby("Department")["Marks"].mean()
```

```python
# Group by Multiple Columns

students.groupby(["Department","Gender"])["Marks"].mean()
```

### 2. agg()

agg() (Aggregate) performs multiple calculations on grouped data simultaneously.

Syntax:

df.groupby("column").agg(functions)

Example:

A bank calculates:

* Average balance
* Maximum balance
* Number of customers

for each branch.

#### AI/ML Usage

Used to generate statistical summaries of grouped features.

```python
students.groupby("Department")["Marks"].agg(
    ["mean","max","min","count"]
)
```

```python
# Multiple Columns
students.groupby("Department").agg({
    "Marks":["mean","max"],
    "Age":"mean"
})
```

### 3. transform()
Definition

transform() performs calculations on each group but returns a result with the same number of rows as the original DataFrame.

Example:

Each employee's salary is compared with the average salary of their department.

#### AI/ML Usage

Used for feature engineering by creating normalized or group-based features.

```python
students["Department_Average"] = students.groupby(
    "Department"
)["Marks"].transform("mean")

print(students)
```

### 4. filter()

filter() removes groups that do not satisfy a specified condition.

Example:

A company keeps only departments whose average sales exceed ₹1,00,000.

#### AI/ML Usage

Useful for removing weak or insignificant groups before analysis.

```python
students.groupby("Department").filter(
    lambda x: x["Marks"].mean() > 90
)
```

### 5. Pivot Tables

A Pivot Table summarizes data by grouping rows and columns while applying aggregation functions.

It provides a spreadsheet-like summary.

Syntax:

pd.pivot_table()

Example:

A company creates a report showing average monthly sales for each branch.

#### AI/ML Usage

Pivot tables are useful for quickly summarizing datasets during EDA.

```python
pd.pivot_table(
    students,
    values="Marks",
    index="Department",
    aggfunc="mean"
)
```

```python
# Multiple Aggregations
pd.pivot_table(
    students,
    values="Marks",
    index="Department",
    aggfunc=["mean","max","min"]
)
```

```python
# Pivot with Rows and Columns
pd.pivot_table(
    students,
    values="Marks",
    index="Department",
    columns="Gender",
    aggfunc="mean"
)
```

### 6. Crosstab

pd.crosstab() creates a frequency table showing how often combinations of categorical values occur.

Syntax:

pd.crosstab(row,column)

Example:

A hospital counts male and female patients in each department.

#### AI/ML Usage

Used to study relationships between categorical variables before model building.

```python
pd.crosstab(
    students["Department"],
    students["Gender"]
)
```

```python
# Crosstab with Totals
pd.crosstab(
    students["Department"],
    students["Gender"],
    margins=True
)
```

```python
# Crosstab with Percentages
pd.crosstab(
    students["Department"],
    students["Gender"],
    normalize=True
)
```
