# Data Inspection

```python
import pandas as pd
```

#### Create a Sample Dataset

```python
students = pd.DataFrame({
    "ID": [101, 102, 103, 104, 105],
    "Name": ["Aishu", "Rahul", "Priya", "John", "Meena"],
    "Age": [22, 23, 21, 24, 22],
    "Department": ["AI", "ML", "DS", "AI", "ML"],
    "Marks": [95, 88, 91, 84, 89]
})

print(students)
```

### 1. head()

head() displays the first five rows of a DataFrame by default.

It is useful for quickly checking whether the dataset has been loaded correctly.

Syntax:

df.head()

We can also specify the number of rows.

df.head(n)

Example:

A data analyst loads a CSV file containing 1 million customer records. Instead of displaying the entire dataset, they use head() to verify that the data has been imported correctly.

#### AI/ML Usage

Used to quickly inspect datasets before preprocessing and model training.

```python
students.head()
```

### 2. tail()

tail() displays the last five rows of a DataFrame.

Example:

A company checks the latest customer transactions by viewing the last few rows of a daily sales dataset.

#### AI/ML Usage

Useful for verifying recently appended records in large datasets.

```python
students.tail(2)
```

### 3. sample()

sample() returns randomly selected rows from a DataFrame.

Syntax:

df.sample(n)

Example:

A teacher randomly selects two students for an oral presentation.

#### AI/ML Usage

Used for random data inspection, testing, and creating sample datasets.

```python
students.sample(2)
```

### 4. info()

info() provides a summary of the DataFrame.

It displays:

* Number of rows
* Number of columns
* Column names
* Data types
* Missing values
* Memory usage

Example:

Before cleaning employee records, an HR analyst checks which columns have missing values and their data types.

#### AI/ML Usage

Used to understand the dataset before feature engineering and preprocessing.

```python
students.info()
```

### 5. describe()

describe() generates statistical summaries of numerical columns.

It provides:

* Count
* Mean
* Standard deviation
* Minimum
* Maximum
* Quartiles
  
Example:

A hospital calculates the average age and maximum age of patients.

#### AI/ML Usage

Used to identify outliers and understand feature distributions.

```python
students.describe()
```

### 6. shape

shape returns the number of rows and columns.

Example:

A company checks whether the downloaded sales dataset contains the expected number of records.

#### AI/ML Usage

Used to verify dataset size before model training.

```python
students.shape
```

Meaning:

5 rows

5 columns

### 7. columns

columns returns the names of all columns.

Example:

An analyst checks available columns before selecting features for analysis.

#### AI/ML Usage

Used for feature selection and column renaming.

```python
students.columns
```

### 8. dtypes

dtypes displays the data type of every column.

Example:

A bank verifies whether transaction amounts are stored as numbers instead of text.

#### AI/ML Usage

Correct data types are required before applying machine learning algorithms.

```python
students.dtypes
```

### 9. memory_usage()

memory_usage() shows how much memory each column occupies.

Example:

A company working with millions of records checks memory usage to optimize storage and improve performance.

#### AI/ML Usage

Used to reduce memory consumption in large datasets.

```python
students.memory_usage()
```

### 10. nunique()

nunique() returns the number of unique values in each column.

Example:

A supermarket checks how many different product categories are available.

#### AI/ML Usage

Helps identify categorical features and low-cardinality columns.

```python
students.nunique()
```

### 11. value_counts()

value_counts() counts how many times each unique value appears in a column.

Example:

An HR department counts how many employees belong to each department.

#### AI/ML Usage

Useful for checking class balance in classification problems.

```python
students["Department"].value_counts()
```
