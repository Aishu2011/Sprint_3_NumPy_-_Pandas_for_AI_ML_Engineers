# NumPy Mathematical Operations

```python
import numpy as np
```

#### sample dataset

```python
scores = np.array([78, 85, 92, 67, 88, 95, 73, 81])

print(scores)
```

### 1. Sum

The Sum function calculates the total of all elements in an array.

It is useful when finding the total marks, total sales, total revenue, or total values in a dataset.

```python
total = np.sum(scores)

print(total)
```

### 2. Mean

The Mean is the average value of all numbers.

It is calculated by dividing the sum of all values by the total number of values.

```python
average = np.mean(scores)

print(average)
```

### 3. Median

The Median is the middle value after arranging the numbers in ascending order.

If there are two middle values, NumPy returns their average.

```python
median = np.median(scores)

print(median)
```

### 4. Variance

Variance measures how far the data points are spread from the mean.

* Small variance → values are close together.
* Large variance → values are widely spread.

```python
variance = np.var(scores)

print(variance)
```

### 5. Standard Deviation

Standard Deviation measures the average distance of data points from the mean.

It is simply the square root of the variance.

```python
std = np.std(scores)

print(std)
```

### 6. Minimum

Returns the smallest value in the array.

```python
minimum = np.min(scores)

print(minimum)
```

### 7. Maximum

Returns the largest value in the array.

```python
maximum = np.max(scores)

print(maximum)
```

### 8. Argmin

Returns the index of the smallest value.

* min() → value
* argmin() → position

```python
print(np.argmin(scores))
```

Because

scores[3] = 67

### 9. Argmax

Returns the index of the largest value.

```python
print(np.argmax(scores))
```

Because

scores[5] = 95

### 10. Percentile

A percentile indicates the value below which a certain percentage of observations fall.

Examples:

* 25th percentile
* 50th percentile (Median)
* 75th percentile

```python
print(np.percentile(scores,25))

print(np.percentile(scores,50))

print(np.percentile(scores,75))
```

### 11. Dot Product

The dot product multiplies corresponding elements of two vectors and then adds the results.

It is one of the most important operations in Machine Learning.

```python
a = np.array([2,3,4])

b = np.array([5,6,7])

print(np.dot(a,b))
```

Calculation

(2×5)+(3×6)+(4×7)

10+18+28=56

### 12. Matrix Multiplication

Matrix multiplication combines two matrices to produce another matrix.

The number of columns in the first matrix must equal the number of rows in the second matrix.

```python
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])
```

#### Method 1

```python
print(np.matmul(A,B))
```

#### Method 2

```python
print(np.matmul(A,B))
```

### 13. Transpose

Transpose converts rows into columns and columns into rows.

```python
A = np.array([[1,2,3],
              [4,5,6]])

print(A.T)
```

### 14. Basic Linear Algebra Operations

Linear Algebra forms the mathematical foundation of Artificial Intelligence and Machine Learning.

#### Addition

```python
A = np.array([[1,2],
              [3,4]])

B = np.array([[5,6],
              [7,8]])

print(A+B)
```

#### Subtraction

```python
print(A-B)
```

#### Scalar Multiplication

```python
print(A*3)
```

#### Matrix Multiplication

```python
print(A@B)
```

#### Determinant

The determinant is a single value that tells us important properties of a square matrix, such as whether it has an inverse.

```python
print(np.linalg.det(A))
```

#### Inverse

The inverse of a matrix is used to solve systems of linear equations.

```python
print(np.linalg.inv(A))
```

#### Rank

The rank of a matrix tells the number of linearly independent rows or columns.

```python
print(np.linalg.matrix_rank(A))
```

#### Eigenvalues and Eigenvectors

Eigenvalues and eigenvectors describe the directions in which a matrix stretches or compresses data.

```python
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues")

print(eigenvalues)

print()

print("Eigenvectors")

print(eigenvectors)
```
