### 1. What is Pandas?

Pandas is an open-source Python library used for working with structured or tabular data. It provides powerful data structures like Series and DataFrame to efficiently store, manipulate, analyze, and visualize data.

Pandas simplifies tasks such as:

* Reading data from files
* Cleaning missing values
* Filtering data
* Sorting data
* Grouping data
* Performing statistical analysis

Pandas is built on top of the NumPy library, making it fast and efficient.

#### Example

Imagine a school has a spreadsheet containing information about 1,000 students.

Instead of manually calculating averages or finding the highest mark, Pandas can perform these operations in just a few lines of code.

```python
import pandas as pd

students = {
    "Name": ["Aishu", "Rahul", "Priya"],
    "Marks": [95, 88, 91]
}

df = pd.DataFrame(students)

print(df)
```

#### AI/ML Usage

Pandas is commonly used to:

* Load datasets
* Clean data
* Handle missing values
* Select features
* Prepare data before training machine learning models


### 2. Why Pandas?

Working with large datasets using Python lists or dictionaries can be difficult and time-consuming.

Pandas provides simple functions to efficiently perform data analysis and manipulation.

Advantages of Pandas:

* Fast and efficient
* Easy to read CSV and Excel files
* Handles missing values
* Supports filtering and sorting
* Performs statistical analysis
* Works well with NumPy and Scikit-learn

#### Example

An online shopping company stores millions of customer orders every day.

Using Pandas, they can easily answer questions like:

* Which product sold the most?
* Which customer spent the highest amount?
* What was today's total revenue?

```python
import pandas as pd

sales = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet"],
    "Sales": [50, 120, 35]
})

print(sales["Sales"].sum())
```

### 3. Series

A Series is a one-dimensional labeled array.

It can store:

* Numbers
* Strings
* Boolean values

Each value has an index.

```python
import pandas as pd

marks = pd.Series([90, 85, 95, 88])
print(marks)
```

#### Custom Index


```python
marks = pd.Series(
    [90,85,95],
    index=["Math","Science","English"]
)

print(marks)
```

#### Example

A student's marks in different subjects.

This is naturally represented as a Series.

#### AI/ML Usage

Series are commonly used for:

* Target variables
* Prediction outputs
* Individual feature columns

### 4. DataFrame

A DataFrame is a two-dimensional table consisting of rows and columns.

It is the most commonly used Pandas data structure.

Each column can have different data types.

```python
import pandas as pd

students = {
    "Name":["Aishu","Rahul","Priya"],
    "Age":[22,23,21],
    "Marks":[95,88,91]
}

df = pd.DataFrame(students)

print(df)
```

#### Example
A company's employee database.

#### AI/ML Usage

Machine learning datasets are usually stored as DataFrames before preprocessing.

### 5. Creating DataFrames

Pandas allows DataFrames to be created from different data sources.

#### A. From Dictionary

```python
import pandas as pd

data = {
    "Name":["Aishu","Rahul"],
    "Age":[22,23]
}

df = pd.DataFrame(data)

print(df)
```

#### B. From List of Lists

```python
data = [
    ["Aishu",22],
    ["Rahul",23]
]

df = pd.DataFrame(
    data,
    columns=["Name","Age"]
)

print(df)
```

#### C. From NumPy Array

```python
import numpy as np

arr = np.array([[1,2],[3,4]])

df = pd.DataFrame(
    arr,
    columns=["A","B"]
)

print(df)
```

### Example

A company stores employee details in different formats (Excel, database, or arrays). Pandas can convert all of them into DataFrames for analysis.

### AI/ML Usage

Most datasets downloaded from Kaggle or company databases are converted into DataFrames before model training.

### 6. Data Types

Each DataFrame column has a specific data type that determines what kind of values it can store.

Common data types include:

### 7. Index

The Index uniquely identifies each row in a DataFrame.

By default, Pandas assigns indexes starting from 0.

```python
import pandas as pd

df = pd.DataFrame({
    "Name":["Aishu","Rahul"],
    "Marks":[95,88]
})

print(df.index)
```

### 9. Reading Data

Reading data means importing data from external files into a Pandas DataFrame so that it can be analyzed and processed.

Pandas supports many file formats such as:

* CSV
* Excel
* JSON
* SQL Database
* Parquet

#### Create a Sample DataFrame

```python
import pandas as pd

students = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Aishu", "Rahul", "Priya", "John"],
    "Age": [22, 23, 21, 24],
    "Marks": [95, 88, 91, 84]
})

print(students)
```

#### Create a CSV File

```python
students.to_csv("students.csv", index=False)

print("CSV file created successfully.")
```

#### Read CSV File

```python
df = pd.read_csv("students.csv")

print(df)
```

#### Create an Excel File

```python
students.to_excel("students.xlsx", index=False)

print("Excel file created successfully.")
```

#### Read Excel File

```python
df = pd.read_excel("students.xlsx")

print(df)
```

#### Create a JSON File

```python
students.to_json("students.json", orient="records", indent=4)

print("JSON file created successfully.")
```

#### Read JSON File

```python
df = pd.read_json("students.json")

print(df)
```

#### AI/ML Usage

Datasets downloaded from Kaggle or company databases are usually stored as CSV or Excel files. Pandas loads these datasets before preprocessing and training machine learning models.

### 10. Writing Data

Writing data means saving a DataFrame into a file so it can be shared, stored, or used later.

Pandas supports writing to:

* CSV
* Excel
* JSON

#### Create a New DataFrame

```python
import pandas as pd

employees = pd.DataFrame({
    "Employee_ID": [1, 2, 3],
    "Name": ["John", "Sara", "David"],
    "Department": ["IT", "HR", "Finance"],
    "Salary": [50000, 45000, 60000]
})

print(employees)
```

#### Write to CSV

```python
employees.to_csv("employees.csv", index=False)

print("employees.csv saved successfully.")
```

#### Write to Excel

```python
employees.to_excel("employees.xlsx", index=False)

print("employees.xlsx saved successfully.")
```

#### Write to JSON

```python
employees.to_json("employees.json", orient="records", indent=4)

print("employees.json saved successfully.")
```

#### Verify the Saved Files

```python
csv_data = pd.read_csv("employees.csv")

excel_data = pd.read_excel("employees.xlsx")

json_data = pd.read_json("employees.json")

print(csv_data)

print()

print(excel_data)

print()

print(json_data)
```

#### AI/ML Usage

After preprocessing a dataset (removing duplicates, filling missing values, encoding categorical columns, etc.), data scientists often save the cleaned dataset before model training.
